```python
import numpy as np
import pandas as pd

#filename = "/Users/josephtrevorrow/Documents/GitHub/HCVA-plus-plus/results/Country/DIR_0_RUN_0_2026-10-20.csv "
#filename = "/home/ia23938/PycharmProjects/HCVA-plus-plus/plots/ESS/Entire PVS Residualsresiduals.csv"
filename = "/home/ia23938/PycharmProjects/HCVA-plus-plus/plots/ESS/PriPs Residualsresiduals.csv"

df = pd.read_csv(filename, converters={'points': pd.eval})
rows_to_plot = ["HCVA++", "HCVA", "SLM", "T", "1", "inf"]
df = df.loc[df['key'].isin(rows_to_plot)]

# For each row, take points, and save it as a filemane
for row in df.iterrows():
    points = row[1]['points']
    filename = str(row[1]['key']) + "_prips.csv"
    df_to_save = pd.DataFrame(points)
    df_to_save.to_csv(filename, index=False, header="Residuals")
```

```python
## Find the standard deviation and variance and save to another file for tracking
# For each row, take points, and save it as a filename
import csv

rows = [["key", "std", "var", "mean", "median"]]
for row in df.iterrows():
    points = row[1]['points']
    temp_std = np.std(points)
    temp_var = np.var(points)
    temp_mean = np.mean(points)
    temp_median = np.median(points)
    rows.append([row[1]['key'], temp_std, temp_var, temp_mean, temp_median])

filename = "ess_just_prips.csv"
with open(filename, 'w', newline='') as csvfile:
    # writing file
    writer = csv.writer(csvfile)
    writer.writerows(rows)
csvfile.close()
```
